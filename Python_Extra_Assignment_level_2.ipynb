8.Print Odd Numbers in the list

```python
arr=[20,10,16,19,25,1,276,188]

for val in arr:
    if val % 2 == 1:
        print(f"{val} is odd")
```

9.Print Even Numbers in the list

```python
arr=[20,10,16,19,25,1,276,188]

for val in arr:
    if val % 2 == 0:
        print(f"{val} is even")
```
