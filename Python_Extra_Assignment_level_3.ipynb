3.Catagory the people by their age like children, adult, citizen, senior citizen...

```python
age=int(input("Enter the age:"))
print("age:",age)
if age < 18:
    print("Children")
elif age < 35:
    print("Adult")
elif age < 50:
    print("Citizen")
elif age < 60:
    print("Senior citizen")
```

4.Find whether given number is positive or negative

```python
num=int(input("Enter a number:"))
print(f"Enter any number:{num}")
if num < 0:
    print(f"{num} is Negative")
else:
    print(f"{num} is Positive")
```

5.Check whether the given number is divisible by 5

```python
num=int(input("Enter a number:"))
print(f"Enter a number to check:{num}")

if num % 5 != 0:
    print(f"{num} is not divisible by 5")
else:
    print(f"{num} is divisible by 5")
```
