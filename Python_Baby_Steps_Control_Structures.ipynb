Body Mass Index

```python
weight=float(input("Enter the weight:"))
print(weight)
if 18.5 <= weight <= 24.9:
    print("Normal")
elif 25 <= weight <= 29.9:
    print("Overweight")
elif 30 <= weight <= 39.9:
    print("Obese")
else:
    print("Morbidly Obese")
```
