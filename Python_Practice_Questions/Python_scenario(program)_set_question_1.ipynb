1.Scenario: A system checks if a user is eligible to vote based on their age.
 Write logic to ask the user for their age and determine if they are eligible to vote based on whether they are 18 or older

```python
class Voting():
    def vote():
        age=int(input("Enter a age: "))
        print(f"Your age is {age}")
        if age >= 18:
            result = "Eligible for voting"
        else:
            result = "Not Eligible for voting"
        return result
```

```python
Voting.vote()
```

2.Scenario: A program processes a list of numbers and needs to find the largest value.
 Write logic to identify and return the largest number from a given list.

```python
class Largest():
    def large():
        lists=[2,1,3,6,4,9,5]
        largest=lists[0]
        
        for item in lists:
            if item > largest:
                largest=item
        return f"The largest element is {largest}"
```

```python
Largest.large()
```

5.Scenario: A text-processing tool reverses a given word or sentence for formatting purposes.
 Write logic to take a word or sentence as input and produce its reversed version.

```python
class Reversed():
    def reverse():
        text=input("Enter a word or sentence:")
        reversed_word=""
        print(f"Entered word or sentence is {text}")
        
        for letter in text:
            reversed_word = letter + reversed_word
        return reversed_word
```

```python
Reversed.reverse()
```

7.Scenario: A retail store offers a 20% discount if a customer’s total order exceeds $100. Write logic to calculate the final amount to be paid after applying the discount

```python
class Discount():
    def discount():
        total_amount=int(input("Enter the amount:"))
        print(f"Entered amount is {total_amount}")
        
        if total_amount >= 100:
            discount_amount = total_amount * (20/100)
            final_amount = total_amount - discount_amount
            return f"The amount to be given is {final_amount}"
        else:
            return "Sorry you need to buy extra things to reach $100 to get the discount"
```

```python
Discount.discount()
```

9.Scenario: A calendar system verifies whether a given year is a leap year based on standard leap year rules.
 Write logic to determine whether a given year is a leap year.

```python
class Leap():
    def leap():
        year=int(input("Enter any year:"))
        print(f"Entered year is {year}")
        
        if year % 4 == 0:
            message = "Leap year"
        if year % 100 == 0:
            message = "Not a Leap year"
        if year % 400 == 0:
            message = "Leap year"
        else:
            message = "Not a Leap year"
        return message
```

```python
Leap.leap()
```

10.Scenario: A program filters out only even numbers from a given list.
 Write logic to extract and return only the even numbers from a list.

```python
class Even():
    def even():
        lists=[1,2,3,4,5,6]
        final_result=list(filter(lambda x:x % 2 == 0,lists))
        return f"The Even numbers in the lists are {final_result}"
```

```python
Even.even()
```
