 #3.Find Even and Odd Numbers in a List
Question: Write a Python program to separate even and odd numbers from the list [12, 17, 19, 24, 35, 40, 56].



```python
arr=[12, 17, 19, 24, 35, 40, 56]
even=[]
odd=[]
for i in arr:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even number is ",even)
print("Odd numbers is ",odd)
```

#4. Reverse a Number
Question: Write a Python program to reverse the number 12345.


```python
num=12345
reversed_number=0

while 0 < num:
    rem = num % 10
    reversed_number = rem + reversed_number * 10
    num//=10
print("Reversed number is",reversed_number)
```

#5. Find the Second Largest Number in a List
Question: Write a Python program to find the second largest number in the list [12, 45, 98, 34, 23].


```python
arr = [12, 45, 98, 34, 23]
large=arr[0]
second_large=arr[0]

for i in arr:
    if i > large:
        second_large=large
        large=i
    elif second_large < i:
        second_large=i
print("The second largest number is",second_large)
```
